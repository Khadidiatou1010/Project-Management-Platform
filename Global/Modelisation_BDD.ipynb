# Analyse fonctionnelle du système

## Contexte

Plusieurs institutions publiques et organisations non gouvernementales souhaitent centraliser la gestion de leurs projets, actuellement dispersée entre fichiers Excel, échanges informels et outils hétérogènes. Cette fragmentation entraîne des retards, des doublons d’information et une visibilité limitée sur l’avancement réel des actions menées.

Face à ces limites, la mise en place d’une plateforme nationale unique apparaît comme une nécessité stratégique. Celle-ci doit offrir un espace collaboratif, sécurisé et structuré permettant de suivre en temps réel les projets, les tâches, les équipes, les documents et les interactions associées.

L’objectif global de cette plateforme est d’harmoniser les pratiques de gestion, de faciliter la coordination entre les différentes organisations et de garantir un pilotage fiable des projets de développement. Elle a vocation à devenir un référentiel commun soutenant la prise de décision et améliorant l’efficacité collective.

--

## Acteurs

### Administrateur
L’administrateur supervise l’ensemble de la plateforme et veille à son bon fonctionnement. Il gère les comptes utilisateurs, leurs rôles ainsi que les équipes. Il dispose d’une visibilité globale sur tous les projets et peut intervenir en cas de problème d’organisation ou de droits d’accès.

### Chef de projet
Le chef de projet est responsable de la planification, de la coordination et du suivi des projets relevant de son équipe. Il crée les projets, organise les tâches et répartit le travail entre les membres. Il suit l’avancement des activités, ajuste les priorités et anime la collaboration au sein de l’équipe.

### Membre d’équipe
Le membre d’équipe participe directement à la réalisation des tâches qui lui sont attribuées. Il met à jour l’état d’avancement de ses tâches, échange avec les autres utilisateurs via les commentaires et consulte les informations relatives aux projets. Son rôle est d’exécuter les activités opérationnelles et de contribuer au bon déroulement des projets.

--

## 3. Fonctionnalités

### Gestion des utilisateurs et des équipes
La plateforme permet de créer des utilisateurs, de définir leurs rôles et de les organiser au sein d’équipes. Chaque utilisateur appartient à au moins une équipe, ce qui structure la collaboration et la répartition du travail. L’administrateur contrôle les accès et veille à ce que chaque équipe dispose des membres nécessaires.

### Gestion des projets
La gestion des projets permet de créer, modifier et suivre les projets d’une équipe. Chaque projet est associé à une seule équipe et contient des informations essentielles telles que le budget, les dates et le statut. Le chef de projet peut y associer plusieurs utilisateurs afin d’assurer le pilotage et la réalisation des activités.

### Gestion des tâches
Le système permet de créer des tâches, de définir leurs priorités et de les assigner à un ou plusieurs membres. Chaque tâche est obligatoirement rattachée à un projet et possède un statut ainsi qu’un pourcentage d’avancement. Une tâche peut également rester non assignée lorsqu’elle est placée en backlog.

### Gestion des activités (logs)
Chaque action réalisée sur une tâche génère automatiquement une activité, telle qu’un commentaire, un changement de statut ou une réassignation. Ces activités permettent de conserver un historique complet et horodaté des actions effectuées. Elles assurent la traçabilité et facilitent le suivi du travail.

### Gestion des documents
La plateforme permet d’ajouter des documents associés aux projets ou aux tâches. Chaque document est caractérisé par un auteur, une date d’upload et un type, et doit être rattaché soit à un projet, soit à une tâche. Cette règle garantit une organisation claire des fichiers et évite les documents non référencés.

### Notifications (optionnel)
Le système peut générer des notifications lorsqu’un événement important survient, tel que la création d’une nouvelle tâche ou l’ajout d’un commentaire. Chaque notification est destinée à un utilisateur spécifique et peut être marquée comme lue ou non lue. Ce mécanisme permet aux membres de rester informés en temps réel.

## Règles de gestion

Les règles de gestion suivantes définissent les principes de fonctionnement de la plateforme et encadrent l’utilisation des différents objets du système.

### Règles liées aux utilisateurs
- **RG1** : Chaque utilisateur est identifié de manière unique dans le système.  
- **RG2** : Un utilisateur possède un nom, un prénom, un email et un rôle.  
- **RG3** : Un utilisateur peut avoir un rôle parmi : Administrateur, Chef de projet ou Membre.  
- **RG4** : La date d’inscription d’un utilisateur est enregistrée.

### Règles liées aux équipes
- **RG5** : Chaque équipe est identifiée de manière unique.  
- **RG6** : Une équipe possède un nom, une description et une date de création.  
- **RG7** : Une équipe regroupe un ou plusieurs utilisateurs.  
- **RG8** : Un utilisateur peut appartenir à une ou plusieurs équipes.  
- **RG9** : L’appartenance d’un utilisateur à une équipe est caractérisée par une date d’affectation.

### Règles liées aux projets
- **RG10** : Chaque projet est identifié de manière unique.  
- **RG11** : Un projet possède un titre, une description, des dates, un budget, un statut et une priorité.  
- **RG12** : Chaque projet est géré par une seule équipe.  
- **RG13** : Une équipe peut gérer un ou plusieurs projets.

### Règles liées aux tâches
- **RG14** : Chaque tâche est identifiée de manière unique.  
- **RG15** : Une tâche possède un titre, une description, des dates prévues, une priorité et un pourcentage d’avancement.  
- **RG16** : Chaque tâche appartient obligatoirement à un seul projet.  
- **RG17** : Un projet peut contenir zéro, une ou plusieurs tâches.

### Règles liées à l’assignation des tâches
- **RG18** : Une tâche peut être assignée à zéro, un ou plusieurs utilisateurs.  
- **RG19** : Un utilisateur peut être assigné à zéro, une ou plusieurs tâches.  
- **RG20** : L’assignation d’un utilisateur à une tâche est caractérisée par une date d’affectation et un rôle dans la tâche.

### Règles liées aux activités (logs)
- **RG21** : Chaque activité est identifiée de manière unique.  
- **RG22** : Une activité possède un contenu, une date/heure et un type.  
- **RG23** : Une activité est réalisée par un seul utilisateur.  
- **RG24** : Une activité est associée à une seule tâche.  
- **RG25** : Une tâche peut générer zéro, une ou plusieurs activités.

### Règles liées aux documents
- **RG26** : Chaque document est identifié de manière unique.  
- **RG27** : Un document possède un nom, un type, une date d’upload et un chemin de fichier.
- **RG28** : Un document est ajouté par un utilisateur.  
- **RG29** : Un document est rattaché soit à un projet, soit à une tâche, mais jamais aux deux simultanément.  
- **RG30** : Un projet ou une tâche peut avoir zéro, un ou plusieurs documents.

### Règles liées aux notifications
- **RG31** : Chaque notification est identifiée de manière unique.  
- **RG32** : Une notification possède un message, un type, un statut et une date/heure.  
- **RG33** : Une notification est reçue par un seul utilisateur.  
- **RG34** : Un utilisateur peut recevoir zéro, une ou plusieurs notifications.

## Exigences non fonctionnelles

### Sécurité
L’accès à la plateforme est contrôlé par des rôles et par l’appartenance aux équipes et aux projets. Les utilisateurs ne peuvent consulter ou modifier que les informations auxquelles ils sont autorisés.

### Traçabilité
Toutes les actions importantes réalisées sur les tâches sont enregistrées dans un historique. Chaque action est associée à un utilisateur et à une date afin d’assurer le suivi et la responsabilité des interventions.

### Collaboration
La plateforme doit permettre à plusieurs utilisateurs de travailler simultanément sur un même projet. Les membres peuvent échanger via des commentaires et suivre l’évolution des tâches en temps réel.

## Uses cases (Scénarios d’utilisation)


Cette partie présente les scénarios d’utilisation (use cases) de la plateforme de gestion des projets, tâches et activités. Les use cases décrivent, de manière concrète, les différentes interactions possibles entre les acteurs du système et la plateforme, en fonction de leurs rôles et responsabilités.
Ils permettent de traduire les besoins fonctionnels en situations réelles d’utilisation, et constituent une base essentielle pour la compréhension du fonctionnement global du système ainsi que pour la modélisation de la base de données.

### Acteur : Administrateur

- UC1 : Créer un utilisateur

L’administrateur crée un nouvel utilisateur en renseignant ses informations personnelles (nom, prénom, email, téléphone) et en lui attribuant un rôle. L’utilisateur est ensuite enregistré dans le système et peut accéder à la plateforme selon ses droits.

- UC2 : Modifier les informations d’un utilisateur

L’administrateur met à jour les informations d’un utilisateur existant (rôle, coordonnées, statut). Les modifications sont immédiatement prises en compte par le système.
-UC3 : Créer une équipe

L’administrateur crée une nouvelle équipe en renseignant son nom et sa description. L’équipe devient disponible pour la gestion des projets.

-UC4 : Ajouter un utilisateur à une équipe

L’administrateur affecte un ou plusieurs utilisateurs à une équipe donnée. La date d’affectation est enregistrée afin de tracer l’appartenance à l’équipe.

-UC5 : Gérer les droits et accès

L’administrateur contrôle les droits d’accès des utilisateurs en fonction de leur rôle et de leur appartenance aux équipes. Il peut intervenir en cas de problème d’accès ou d’organisation.

### Acteur : Chef de projet

- UC6 : Créer un projet

Le chef de projet crée un nouveau projet en renseignant le titre, la description, le budget, les dates prévues, la priorité et le statut. Le projet est automatiquement rattaché à une équipe.

- UC7 : Modifier un projet

Le chef de projet met à jour les informations d’un projet existant (statut, dates, budget ou priorité).
Les changements sont visibles par les membres autorisés.

- UC8 : Ajouter des membres à un projet

Le chef de projet associe un ou plusieurs utilisateurs à un projet afin de constituer l’équipe projet.
Les utilisateurs ajoutés peuvent consulter et intervenir sur le projet.

- UC9 : Créer une tâche

Le chef de projet crée une tâche liée à un projet en renseignant son titre, sa description, ses dates prévues et sa priorité. La tâche est enregistrée avec un statut initial.

- UC10 : Assigner une tâche

Le chef de projet assigne une tâche à un ou plusieurs utilisateurs. Une tâche peut également rester non assignée lorsqu’elle est placée en backlog.

- UC11 : Modifier le statut d’une tâche

Le chef de projet modifie le statut d’une tâche (à faire, en cours, bloquée, terminée).
Cette action génère automatiquement une activité.

- UC12 : Suivre l’avancement du projet

Le chef de projet consulte l’avancement global du projet à travers les tâches et leurs pourcentages d’avancement. Il peut ajuster les priorités si nécessaire.


### Membre d’équipe

- UC13 : Consulter les projets

Le membre d’équipe consulte la liste des projets auxquels il est associé. Il accède aux informations générales et à l’état d’avancement.

- UC14 : Consulter les tâches

Le membre d’équipe consulte les tâches qui lui sont assignées ainsi que celles du projet. Il peut visualiser leur statut et leurs échéances.

- UC15 : Mettre à jour l’avancement d’une tâche

Le membre d’équipe met à jour le pourcentage d’avancement et le statut d’une tâche. Cette mise à jour est enregistrée et génère une activité.

- UC16 : Ajouter un commentaire

Le membre d’équipe ajoute un commentaire sur une tâche afin d’échanger avec les autres membres.
Le commentaire est enregistré dans l’historique des activités.

- UC17 : Ajouter un document

Un utilisateur ajoute un document à un projet ou à une tâche. Le document est enregistré avec son auteur et sa date d’upload.

### Use cases transverses (système)

- UC18 : Génération d’une activité (log)

Le système génère automatiquement une activité lors de toute action importante sur une tâche (commentaire, changement de statut, ajout de document). Chaque activité est horodatée et associée à un utilisateur.

- UC19 : Envoi d’une notification

Le système envoie une notification à un utilisateur lors d’un événement important (création de tâche, commentaire, changement de statut). La notification peut être marquée comme lue ou non lue.

- UC20 : Consultation de l’historique

Un utilisateur consulte l’historique des activités d’une tâche afin de suivre les actions réalisées et les échanges associés.


# Modèle Conceptuel de Données (MCD)

Le **MCD (Modèle Conceptuel de Données)** est la représentation **métier** du système : il décrit *ce qu’on veut gérer* (entités comme Utilisateur, Projet, Tâche, Document…) et *comment ces éléments sont liés* (associations) avec leurs **cardinalités** (1..N, 0..N, etc.). À ce stade, on ne parle pas de SQL : l’objectif est de valider que le modèle colle à la réalité du besoin.

# Modèle Logique de Données (MLD)

Le **MLD (Modèle Logique de Données)** est la **traduction relationnelle** du MCD : on transforme les entités/associations en **tables**, on fixe les **clés primaires (PK)**, les **clés étrangères (FK)**, et on crée les **tables d’association** pour les relations N–N (ex. `ASSIGNER` pour “utilisateur ↔ tâche”). Le MLD explique donc *comment* le modèle métier se mappe dans une base relationnelle, mais reste encore relativement indépendant d’un SGBD particulier.

# Modèle Physique de Données (MPD)

Dans cete section, nous présentons le Modèle Physique de Données (MPD) en tableaux (colonnes, types, clés, contraintes).

Le **MPD (Modèle Physique de Données)** est la version **technique prête à être implémentée** dans un SGBD précis (MySQL, PostgreSQL…). Il reprend le MLD et ajoute tout ce qui est nécessaire pour créer réellement la base : **types SQL** (`INT`, `VARCHAR`, `DATE`, `DATETIME`…), **contraintes** (`NOT NULL`, `UNIQUE`, `CHECK`), règles de cohérence (ex. avancement entre 0 et 100), éventuellement **index** et conventions de stockage. Autrement dit, le MPD dit exactement “comment la base sera stockée” dans un environnement réel.

# Contraintes d’intégrité

Cette partie liste les contraintes d’intégrité du MPD. Les tableaux ci-dessous sont conçus pour un affichage stable (GitHub, VS Code, Obsidian, etc.).

## Définition

Les contraintes d’intégrité sont des règles appliquées par le SGBD pour empêcher l’enregistrement de données incohérentes. Elles couvrent l’intégrité d’entité (chaque ligne est unique via une clé primaire), l’intégrité référentielle (les clés étrangères ne peuvent pointer que vers des enregistrements existants), l’intégrité de domaine (valeurs obligatoires, unicité, plages/ensembles autorisés) et l’intégrité métier (règles propres au contexte, comme la cohérence des dates ou l’exclusivité « projet OU tâche » pour un document). En cas de violation, l’insertion ou la mise à jour est refusée.

SELECT e.nom_equipe, COUNT(a.id_utilisateur) as nombre_membres

## 10 requêtes avancées

- 21. Trouver les projets avec plus de 5 tâches
SELECT p.titre_projet, COUNT(t.id_tache) as nombre_taches
FROM PROJET p
LEFT JOIN TACHE t ON p.id_projet = t.id_projet
GROUP BY p.id_projet, p.titre_projet
HAVING COUNT(t.id_tache) > 5;

- 22. Afficher les équipes ayant un budget total supérieur à 5 000 000 FCFA
SELECT e.nom_equipe, SUM(p.budget_alloue) as budget_total
FROM EQUIPE e
INNER JOIN PROJET p ON e.id_equipe = p.id_equipe
GROUP BY e.id_equipe, e.nom_equipe
HAVING SUM(p.budget_alloue) > 5000000;

- 23. Lister les utilisateurs qui n'ont aucune tâche assignée
SELECT u.id_utilisateur, u.nom_utilisateur, u.prenom_utilisateur
FROM UTILISATEUR u
WHERE u.id_utilisateur NOT IN (
    SELECT DISTINCT id_utilisateur FROM ASSIGNER
);

- 24. Trouver les projets sans aucune tâche
SELECT p.id_projet, p.titre_projet, p.statut
FROM PROJET p
WHERE p.id_projet NOT IN (
    SELECT DISTINCT id_projet FROM TACHE
);

- 25. Afficher les tâches avec leur projet, équipe et utilisateurs assignés
SELECT t.titre_tache, t.statut, p.titre_projet, e.nom_equipe, 
       u.nom_utilisateur, u.prenom_utilisateur
FROM TACHE t
INNER JOIN PROJET p ON t.id_projet = p.id_projet
LEFT JOIN EQUIPE e ON p.id_equipe = e.id_equipe
LEFT JOIN ASSIGNER a ON t.id_tache = a.id_tache
LEFT JOIN UTILISATEUR u ON a.id_utilisateur = u.id_utilisateur;

- 26. Calculer le taux d'avancement global par équipe
SELECT e.nom_equipe, 
       AVG(t.pourcentage_avancement) as taux_avancement_moyen,
       COUNT(t.id_tache) as total_taches
FROM EQUIPE e
INNER JOIN PROJET p ON e.id_equipe = p.id_equipe
INNER JOIN TACHE t ON p.id_projet = t.id_projet
GROUP BY e.id_equipe, e.nom_equipe;

- 27. Trouver les utilisateurs les plus actifs (avec le plus de tâches)
SELECT u.nom_utilisateur, u.prenom_utilisateur, COUNT(a.id_tache) as nb_taches
FROM UTILISATEUR u
INNER JOIN ASSIGNER a ON u.id_utilisateur = a.id_utilisateur
GROUP BY u.id_utilisateur, u.nom_utilisateur, u.prenom_utilisateur
ORDER BY nb_taches DESC
LIMIT 5;

- 28. Lister les projets en retard (date fin prévue dépassée et statut non terminé)
SELECT p.titre_projet, p.date_fin_prevue, p.statut, e.nom_equipe
FROM PROJET p
LEFT JOIN EQUIPE e ON p.id_equipe = e.id_equipe
WHERE p.date_fin_prevue < CURDATE() AND p.statut != 'Terminé';

- 29. Afficher le nombre de documents par type et par projet
SELECT p.titre_projet, d.type_fichier, COUNT(d.id_document) as nombre_documents
FROM PROJET p
LEFT JOIN DOCUMENT d ON p.id_projet = d.id_projet
GROUP BY p.id_projet, p.titre_projet, d.type_fichier
ORDER BY p.titre_projet, nombre_documents DESC;

- 30. Statistiques complètes par équipe (membres, projets, tâches, budget)
SELECT 
    e.nom_equipe,
    COUNT(DISTINCT a.id_utilisateur) as nb_membres,
    COUNT(DISTINCT p.id_projet) as nb_projets,
    COUNT(DISTINCT t.id_tache) as nb_taches,
    COALESCE(SUM(p.budget_alloue), 0) as budget_total
FROM EQUIPE e
LEFT JOIN APPARTENIR a ON e.id_equipe = a.id_equipe
LEFT JOIN PROJET p ON e.id_equipe = p.id_equipe
LEFT JOIN TACHE t ON p.id_projet = t.id_projet
GROUP BY e.id_equipe, e.nom_equipe;

# Proposition d'amélioration de ProjectFlow


Bien que le modèle actuel couvre les fonctionnalités fondamentales de gestion de projet, plusieurs axes d'amélioration peuvent être envisagés pour accroître la performance et l'ergonomie de la plateforme :

- **Pilotage Visuel et Agilité :** L'intégration d'un système de **Kanban digital** permettrait une gestion dynamique des flux de travail, offrant aux utilisateurs une vue globale et intuitive de l'avancement des tâches en temps réel.
- **Analyse Prédictive et IA :** L'ajout d'une couche d'**intelligence artificielle** pourrait assister les chefs de projet dans la suggestion automatique de plannings et la prévision des retards potentiels en se basant sur l'historique des activités.
- **Gouvernance et Sécurité :** L'implémentation d'un module de **gestion des risques** permettrait d'identifier et de suivre les points critiques de chaque projet.De plus, un **système de validation** hiérarchique pour l'approbation des livrables renforcerait le contrôle qualité au sein des équipes.
- **Évaluation et Reporting :** Le modèle pourrait être étendu pour inclure l'**évaluation des membres** afin de mesurer la performance individuelle et collective , ainsi que la génération automatique de **rapports PDF** synthétisant l'état de santé financier et opérationnel des projets nationaux.
